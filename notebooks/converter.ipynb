### SIMPLECNN - MNIST Converter


```python
import torch
from models.simplecnn import SimpleCNN

pth_path = "../record/20260629_182416_badnet_attack_badnet_6MrG/mnist_clean_model.pth"   
pt_path  = "../record/20260629_182416_badnet_attack_badnet_6MrG/mnist_clean_model.pt"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# Load checkpoint safely on CPU first (recommended)
checkpoint = torch.load(pth_path, map_location="cpu")
# Initialize model
model = SimpleCNN()
# Load correct key
model.load_state_dict(checkpoint)
model.to(device)

# Save full model object (.pt)
torch.save(model, pt_path)
print(f"[convert] Converted {pth_path} -> {pt_path}")
```
