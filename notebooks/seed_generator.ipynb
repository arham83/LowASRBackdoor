```python
from torchvision import datasets, transforms
from collections import defaultdict
import torch
import pickle

def create_seed_data(samples_per_class=5):
    transform = transforms.ToTensor()

    dataset = datasets.MNIST(
        root="../data",
        train=False,
        download=True,
        transform=transform
    )

    images = []
    labels = []
    counts = defaultdict(int)

    for img, label in dataset:
        if counts[label] < samples_per_class:
            images.append(img)
            labels.append(label)
            counts[label] += 1

        if all(counts[c] >= samples_per_class for c in range(10)):
            break

    seed_images = torch.stack(images)
    seed_labels = torch.tensor(labels)

    return seed_images, seed_labels


samples_per_class = 5
save_path = "../seeds/mnist_seed_data.pkl"
images, labels = create_seed_data(samples_per_class)

with open(save_path, "wb") as f:
    pickle.dump((images.numpy(), labels.numpy()), f)

print(f"Saved seed data to {save_path}")
print(f"Images shape: {images.shape}")
print(f"Labels shape: {labels.shape}")
```
